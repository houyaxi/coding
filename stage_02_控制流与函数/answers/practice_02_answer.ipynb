## 题 2 参考答案：成绩评级函数

```python
def grade(score):
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 70:
        return "中等"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(grade(95))    # 优秀
print(grade(73))    # 中等
print(grade(45))    # 不及格
```

## 题 3 参考答案：累加求和函数

```python
def sum_to(n):
    total = 0
    i = 1
    while i <= n:
        total = total + i
        i = i + 1
    return total

print(sum_to(100))   # 5050
```

## 题 4 参考答案：统计大于平均值

也可以用列表推导式简化最后一步：
greater = [x for x in nums if x > avg]

```python
import random

nums = [random.randint(1, 100) for _ in range(20)]
print(f"生成的列表：{nums}")

avg = sum(nums) / len(nums)
print(f"平均值：{round(avg, 2)}")

count = 0
for x in nums:
    if x > avg:
        count = count + 1

print(f"大于平均值的个数：{count}")
```

## 题 5 参考答案：猜数字小游戏

```python
import random

secret = random.randint(1, 50)
times = 0

while True:
    guess = int(input("请猜一个 1-50 的数字："))
    times = times + 1
    if guess > secret:
        print("大了")
    elif guess < secret:
        print("小了")
    else:
        print(f"猜对了！共猜了{times}次")
        break
```
