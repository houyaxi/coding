# 第二阶段第二课：for 循环与 while 循环

循环让程序重复做同一件事，是消除重复代码的核心武器。

## 1. for 循环遍历列表

for 变量 in 列表：依次取出列表里的每个元素，执行一次缩进的代码块。

```python
hobbies = ["编程", "跑步", "看电影"]

for h in hobbies:
    print(f"我喜欢{h}")

print("循环结束")
```

## 2. range()：生成数字序列

range(5) 生成 0,1,2,3,4（不含 5）；range(1,6) 生成 1 到 5；range(1,10,2) 表示从 1 到 9 步长 2。

```python
# 从 0 到 4
for i in range(5):
    print(i)

print("---")

# 从 1 到 5
for i in range(1, 6):
    print(i)

print("---")

# 步长 2：1, 3, 5, 7, 9
for i in range(1, 10, 2):
    print(i)
```

## 3. 用 for 做累加

在循环外先准备一个变量，循环里不断更新它，这是最常用的模式。

```python
total = 0

for i in range(1, 101):
    total = total + i

print(f"1 到 100 的和是：{total}")
```

## 4. while 循环

while 后面跟条件，条件为 True 就一直循环，直到条件变 False。注意：一定要有让条件变化的语句，否则会死循环。

```python
count = 1

while count <= 5:
    print(f"第{count}次")
    count = count + 1   # 这一行让循环能结束

print("循环结束")
```

## 5. break 和 continue

- break：立刻跳出整个循环
- continue：跳过本次，进入下一次循环

```python
# break 示例：找到第一个能被 7 整除的数就停
for i in range(1, 100):
    if i % 7 == 0:
        print(f"第一个能被7整除的是{i}")
        break
```

```python
# continue 示例：只打印奇数
for i in range(1, 11):
    if i % 2 == 0:
        continue        # 偶数直接跳过
    print(i)
```

## 6. 用循环改写重复代码

第二课随机密码那题你写了 6 遍 random.choice，现在一行循环就搞定：

```python
import random
chars = "abcdefghijklmnopqrstuvwxyz0123456789"
password = ""

for _ in range(6):
    password = password + random.choice(chars)

print(f"生成的密码：{password}")
```
