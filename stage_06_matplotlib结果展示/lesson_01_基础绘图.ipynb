# 第六阶段第一课：matplotlib 基础绘图

matplotlib 是 Python 最基础的绘图库。核心思路：先画图对象，再逐项设置，最后 plt.show() 显示。从上往下一块一块运行。

## 0. 让图直接显示在 notebook 里

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## 1. 折线图：plot

适合看趋势，比如随时间的变化。

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 4, 8, 9])

plt.plot(x, y)
plt.show()
```

## 2. 柱状图：bar

适合对比不同类别的数值。

```python
names = ["语文", "数学", "英语"]
scores = [85, 92, 78]

plt.bar(names, scores)
plt.show()
```

## 3. 散点图：scatter

适合看两个变量之间的关系。

```python
np.random.seed(1)
x = np.random.randn(50)
y = np.random.randn(50)

plt.scatter(x, y)
plt.show()
```

## 4. 直方图：hist

适合看数据的分布。

```python
data = np.random.randn(1000)
plt.hist(data, bins=30)
plt.show()
```
