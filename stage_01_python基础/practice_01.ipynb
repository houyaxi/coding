## 题 4：面积计算

用 math.pi 计算一个半径 r = 12 的圆的面积，再计算半径翻倍后的面积，打印两者之比。

```python
# 你的代码写在这里：
import math
r1=12
s1 = r1*r1*math.pi

r2=r1*2
s2=r2*r2*math.pi

ratio= s1/s2

print(s1, s2, ratio)
```
